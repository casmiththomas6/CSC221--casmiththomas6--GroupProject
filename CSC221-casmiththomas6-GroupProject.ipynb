This web-scraping code is collecting data from a table about English personal pronouns, such as the subjective and objective cases and the
singular and plurals. The table is sourced from Wikipedia.

* This code is to import libraries that will alow further code to parse HTML, open URLS, etc.

```python
# Code to avoid potential SLL issues
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

# Import and open library used to open URL's and import BeautifulSoup/CSV 
from urllib.request import Request, urlopen
from bs4 import BeautifulSoup
import csv
# Specify the url  
url = "https://en.wikipedia.org/wiki/English_language#Determiners"
headers = {'User-Agent': 'Mozilla/5.0'}
 
# Open the url using urllib.request and put the HTML into the page variable
# page = urllib.request.urlopen(url)
req = Request(url, headers=headers)
page = urlopen(req).read()
 
# Parse the HTML from the URL into the BeautifulSoup parse tree format
soup = BeautifulSoup(page, "lxml")

```

* This code finds all the tables on the page and filters the tables to find the right table
* The code also includes a check to make sure if a table is found or not
* If the table is found, the output will print "Table Found!", and it will print "Table not Found!" if not

```python
tables = soup.find_all("table", {"class": "wikitable"})

target_table = None

# Locate the table that contains "Personal pronouns"
for table in tables:
    if "personal pronouns" in table.get_text().lower():
        target_table = table
        break

# Safety check (if it can't find the table)
if target_table is None:
    print("Table not found!")
else:
    rows = target_table.find_all("tr")

    data = []

    for row in rows:
        cols = row.find_all(["th", "td"])
        cols = [col.get_text(strip=True) for col in cols]
        if cols:
            data.append(cols)
    print("Table Found!")
```

* This final bit of code is to write to a CSV file all the data from the table
* If successful, the code will print "Data successfully written to CSC221-webscraping-data.csv" and create the file in the current directory

```python
# Write to CSV
filename = "CSC221-webscrape-data.csv"

with open(filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(data)

print(f"Data successfully written to {filename}")
```
